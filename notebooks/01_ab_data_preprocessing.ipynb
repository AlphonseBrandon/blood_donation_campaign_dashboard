```python
# Drop the "Sélectionner_"ok"_pour_envoyer" column from the "Volontaire" sheet
interim["Volontaire"].drop(columns=['Sélectionner_"ok"_pour_envoyer'], inplace=True)

# Verify the column has been dropped
print(interim["Volontaire"].columns.tolist())
```

```python
# Convert "Taux_d’hémoglobine" to float
interim["Volontaire"]["Taux_d’hémoglobine"] = pd.to_numeric(interim["Volontaire"]["Taux_d’hémoglobine"], errors='coerce')

# Calculate the mean value
mean_of_taux_hemoglobine = interim["Volontaire"]["Taux_d’hémoglobine"].mean().round(1)

# Fill missing values with the mean value
interim["Volontaire"]["Taux_d’hémoglobine"].fillna(mean_of_taux_hemoglobine, inplace=True)

# Verify the changes
print(interim["Volontaire"]["Taux_d’hémoglobine"].head())
print("Number of missing values in 'Taux_d’hémoglobine' after replacement:", interim["Volontaire"]["Taux_d’hémoglobine"].isnull().sum())
```

```python
# Check for columns with missing values in the "Volontaire" sheet
missing_values = interim["Volontaire"].isnull().sum()
columns_with_missing_values = missing_values[missing_values > 0]
print("Columns with missing values:", columns_with_missing_values)
```

```python
# Remove strings ending with "(Non précisé )" in the "Arrondissement_de_résidence" column
interim["Volontaire"]["Arrondissement_de_résidence"] = interim["Volontaire"]["Arrondissement_de_résidence"].str.replace(r'\s*\(Non précisé \)$', '', regex=True)

# Verify the changes
print(interim["Volontaire"]["Arrondissement_de_résidence"].unique())
```

Feature engineering

```python
# Calculate IMC and add it as a new column
interim["Volontaire"]["IMC"] = (interim["Volontaire"]["Poids"] / (interim["Volontaire"]["Taille"] / 100) ** 2).round(2)

# Verify the changes
print(interim["Volontaire"][["Poids", "Taille", "IMC"]].head())
```

```python
# Fréquence_de_don
interim["Volontaire"]["Fréquence_de_don"] = interim["Volontaire"]["A-t-il_(elle)_déjà_donné_le_sang"].apply(lambda x: 1 if x == 'Oui' else 0)
```

```python
# Drop the "Si_oui_preciser_la_date_du_dernier_don" column from the "Volontaire" sheet
interim["Volontaire"].drop(columns=["Si_oui_preciser_la_date_du_dernier_don"], inplace=True)

# Verify the column has been dropped
print(interim["Volontaire"].columns.tolist())
```

```python
# Groupe_d'âge
bins = [18, 25, 35, 45, 55, 65, 75, 85, 95]
labels = ['18-25', '26-35', '36-45', '46-55', '56-65', '66-75', '76-85', '86-95']
interim["Volontaire"]["Groupe_d'âge"] = pd.cut(interim["Volontaire"]["Age"], bins=bins, labels=labels, right=False)

# Catégorie_d'IMC
IMC_bins = [0, 18.5, 24.9, 29.9, 39.9, 100]
IMC_labels = ['Insuffisance pondérale', 'Poids normal', 'Surpoids', 'Obèse', 'Obésité sévère']
interim["Volontaire"]["Catégorie_d'IMC"] = pd.cut(interim["Volontaire"]["IMC"], bins=IMC_bins, labels=IMC_labels, right=False)


# Statut_d'éligibilité
interim["Volontaire"]["Statut_d'éligibilité"] = interim["Volontaire"]["ÉLIGIBILITÉ_AU_DON"].apply(lambda x: 1 if x == 'Eligible' else 0)

# Catégorie_de_niveau_d'hémoglobine
hemoglobin_bins = [0, 12, 16, 20]
hemoglobin_labels = ['Bas', 'Normal', 'Élevé']
interim["Volontaire"]["Catégorie_de_niveau_d'hémoglobine"] = pd.cut(interim["Volontaire"]["Taux_d’hémoglobine"], bins=hemoglobin_bins, labels=hemoglobin_labels, right=False)

# Verify the changes
interim["Volontaire"].head()
```

```python
# Define the new column order
new_column_order = [
    'ID', 'Age','Groupe_d\'âge', 'Taille', 'Poids', 'IMC', 'Catégorie_d\'IMC', 'Fréquence_de_don', 'Taux_d’hémoglobine', 'Catégorie_de_niveau_d\'hémoglobine', 'Statut_d\'éligibilité', 'ÉLIGIBILITÉ_AU_DON', 'Genre', 'Niveau_d\'etude', 'Situation_Matrimoniale_(SM)', 'Profession', 
    'Arrondissement_de_résidence', 'Quartier_de_Résidence', 'Nationalité', 'Religion', 
    'A-t-il_(elle)_déjà_donné_le_sang',  
    'Raison_indisponibilité__[Est_sous_anti-biothérapie__]', 'Raison_indisponibilité__[Taux_d’hémoglobine_bas_]', 
    'Raison_indisponibilité__[date_de_dernier_Don_<_3_mois_]', 'Raison_indisponibilité__[IST_récente_(Exclu_VIH,_Hbs,_Hcv)]', 
    'Raison_de_l’indisponibilité_de_la_femme_[La_DDR_est_mauvais_si_<14_jour_avant_le_don]', 
    'Raison_de_l’indisponibilité_de_la_femme_[Allaitement_]', 'Raison_de_l’indisponibilité_de_la_femme_[A_accoucher_ces_6_derniers_mois__]', 
    'Raison_de_l’indisponibilité_de_la_femme_[Interruption_de_grossesse__ces_06_derniers_mois]', 
    'Raison_de_l’indisponibilité_de_la_femme_[est_enceinte_]', 'Raison_de_non-eligibilité_totale__[Antécédent_de_transfusion]', 
    'Raison_de_non-eligibilité_totale__[Porteur(HIV,hbs,hcv)]', 'Raison_de_non-eligibilité_totale__[Opéré]', 
    'Raison_de_non-eligibilité_totale__[Drepanocytaire]', 'Raison_de_non-eligibilité_totale__[Diabétique]', 
    'Raison_de_non-eligibilité_totale__[Hypertendus]', 'Raison_de_non-eligibilité_totale__[Asthmatiques]', 
    'Raison_de_non-eligibilité_totale__[Cardiaque]', 'Raison_de_non-eligibilité_totale__[Tatoué]', 
    'Raison_de_non-eligibilité_totale__[Scarifié]',
]

# Reassign the dataframe with the new column order
interim["Volontaire"] = interim["Volontaire"][new_column_order]

# Verify the changes
interim["Volontaire"].head()
```

```python
missing_values = interim["Volontaire"].isnull().sum()
columns_with_missing_values = missing_values[missing_values > 0]
print("Columns with missing values:", columns_with_missing_values)
```

```python
# Fill missing values in 'Groupe_d'âge' with the most frequent value
most_frequent_groupe_d_age = interim["Volontaire"]["Groupe_d'âge"].mode()[0]
interim["Volontaire"]["Groupe_d'âge"].fillna(most_frequent_groupe_d_age, inplace=True)

# Fill missing values in 'Catégorie_de_niveau_d'hémoglobine' with the most frequent value
most_frequent_hemoglobin_category = interim["Volontaire"]["Catégorie_de_niveau_d'hémoglobine"].mode()[0]
interim["Volontaire"]["Catégorie_de_niveau_d'hémoglobine"].fillna(most_frequent_hemoglobin_category, inplace=True)

# Verify the changes
missing_values = interim["Volontaire"].isnull().sum()
columns_with_missing_values = missing_values[missing_values > 0]
print("Columns with missing values after handling:", columns_with_missing_values)
```
